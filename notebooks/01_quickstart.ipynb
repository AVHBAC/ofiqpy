# ofiqpy — Quickstart

Assess a face image against OFIQ's ISO/IEC 29794-5 quality components.

> Set `OFIQPY_OFIQ_DATA` to your OFIQ `data/` directory before running.


```python
from ofiqpy import assess
from ofiqpy.config import default_data_root
```

## Assess one image


```python
IMAGE = default_data_root() / "tests" / "images" / "r-01-frontal.png"
scores = assess(IMAGE)
for name, (raw, scalar) in sorted(scores.items()):
    print(f"{name:30} {scalar:3.0f}   (raw {raw:.4f})")
```

## The unified quality score


```python
raw, scalar = scores["UnifiedQualityScore"]
print(f"Unified quality: {scalar}/100  (MagFace magnitude {raw:.2f})")
```

## A quality bar chart


```python
import matplotlib.pyplot as plt

names = sorted(scores)
vals = [scores[n][1] for n in names]
plt.figure(figsize=(8, 7))
plt.barh(names, vals)
plt.xlabel("scalar (0-100)")
plt.xlim(0, 100)
plt.tight_layout()
plt.show()
```
