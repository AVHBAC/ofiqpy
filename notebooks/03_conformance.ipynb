# ofiqpy: Conformance vs live OFIQ

Run the strict 28-image BSI gate. This requires the pinned OFIQ v1.1.0 build and
downloaded canonical data profile.


```python
import os
from pathlib import Path

from ofiqpy.conformance import run_conformance
```

```python
OFIQ_ROOT = Path(os.environ["OFIQPY_OFIQ_ROOT"])
DATA_ROOT = Path(os.environ["OFIQPY_OFIQ_DATA"])
SOURCE_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
```

```python
report = run_conformance(
    ofiq_root=OFIQ_ROOT,
    image_dir=DATA_ROOT / "tests" / "images",
    expected_count=28,
    scalar_tolerance=0.0,
    source_root=SOURCE_ROOT,
)
report.to_dict()
```

The report is complete only when every image/component/status observation is present.
