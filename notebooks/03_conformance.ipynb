# ofiqpy — Conformance vs live OFIQ

Reproduce the ISO Annex A ±1 conformance check. Requires a built `OFIQSampleApp`
(`OFIQPY_OFIQ_ROOT`) and a directory of real face images.


```python
import sys; sys.path.insert(0, '../tests')
from verify_ofiq import run_ofiq, gate, print_report
from ofiqpy.output import OFIQ_ORDER
from ofiqpy.config import OFIQConfig
from ofiqpy.pipeline import OFIQPipeline
from ofiqpy.measures.core import Measures
import cv2, glob
```

```python
IMAGES = sorted(glob.glob('/path/to/images/*.jpg'))[:50]  # <- change
cfg = OFIQConfig(); pipe = OFIQPipeline(cfg); meas = Measures(cfg)
port = {}
for p in IMAGES:
    s = pipe.process(cv2.imread(p))
    if s.bbox is not None:
        port[p.split('/')[-1]] = meas.compute_scalars(s)
```

```python
ofiq_df = run_ofiq([__import__('pathlib').Path(p) for p in IMAGES])
report = gate(port, ofiq_df, list(OFIQ_ORDER), tol=1.0)
print_report(report)
```

A component is **CONFORMANT** when `|port - OFIQ| <= 1` on every image (ISO Annex A.2).
