# ofiqpy — Pipeline internals

Inspect the shared preprocessing products OFIQ computes once per image: the aligned
face, landmarks, parsing / occlusion masks, and head pose.


```python
import cv2
import matplotlib.pyplot as plt

from ofiqpy.config import OFIQConfig, default_data_root
from ofiqpy.pipeline import OFIQPipeline

cfg = OFIQConfig()
pipe = OFIQPipeline(cfg)
```

```python
IMAGE = default_data_root() / "tests" / "images" / "r-01-frontal.png"
s = pipe.process(cv2.imread(str(IMAGE)))
print("bbox:", s.bbox)
print(f"pose  yaw={s.yaw:.1f} pitch={s.pitch:.1f} roll={s.roll:.1f}")
print("landmarks:", s.landmarks.shape, " aligned face:", s.aligned_face.shape)
```

## Aligned face + landmarks


```python
rgb = cv2.cvtColor(s.aligned_face, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(5, 5))
plt.imshow(rgb)
al = s.aligned_landmarks
plt.scatter(al[:, 0], al[:, 1], s=6, c="lime")
plt.title("aligned face + 98 landmarks")
plt.show()
```

## Parsing + occlusion + landmarked region


```python
fig, ax = plt.subplots(1, 3, figsize=(12, 4))
ax[0].imshow(s.parsing, cmap="tab20")
ax[0].set_title("BiSeNet parsing (400x400)")
ax[1].imshow(s.occlusion_mask, cmap="gray")
ax[1].set_title("occlusion (1=visible)")
ax[2].imshow(s.landmarked_region, cmap="gray")
ax[2].set_title("landmarked region")
plt.tight_layout()
plt.show()
```
